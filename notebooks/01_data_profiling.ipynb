# 01 — Data Profiling
Load raw data, inspect schema, report missing values, and run initial quality checks.

```
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_cleaning import (
    load_orders, load_inventory, load_routes,
    clean_orders, clean_inventory, clean_routes, save_processed
)

sns.set_theme(style='whitegrid')
%matplotlib inline
```

## 1.1 Load raw data

```
orders    = load_orders()
inventory = load_inventory()
routes    = load_routes()

print("Orders shape:   ", orders.shape)
print("Inventory shape:", inventory.shape)
print("Routes shape:   ", routes.shape)
```

## 1.2 Schema & dtypes

```
print("\n── Orders dtypes ──")
print(orders.dtypes)
print("\n── Inventory dtypes ──")
print(inventory.dtypes)
print("\n── Routes dtypes ──")
print(routes.dtypes)
```

## 1.3 Missing value rates

```
for name, df in [('Orders', orders), ('Inventory', inventory), ('Routes', routes)]:
    miss = df.isna().mean().sort_values(ascending=False)
    miss = miss[miss > 0]
    print(f"\n[{name}] Missing value rates:")
    print(miss if not miss.empty else "  No missing values.")
```

## 1.4 Clean & save

```
orders_clean    = clean_orders(orders)
inventory_clean = clean_inventory(inventory)
routes_clean    = clean_routes(routes)

save_processed(orders_clean,    'orders_clean.csv')
save_processed(inventory_clean, 'inventory_clean.csv')
save_processed(routes_clean,    'routes_clean.csv')
print("\nAll cleaned files saved to data/processed/")
```
